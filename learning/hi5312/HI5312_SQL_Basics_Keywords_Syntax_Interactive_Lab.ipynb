## 2. SQL is a language for asking questions of data

A basic SQL statement can read almost like English:

```sql
SELECT first_name, last_name
FROM patients;
```

Say it aloud:

> **Select first name and last name from the patients table.**

### The core beginner pattern

```sql
SELECT ...
FROM ...
WHERE ...
ORDER BY ...
LIMIT ...;
```

Each clause has one job:

| SQL keyword | Question it answers |
|---|---|
| `SELECT` | What columns do I want? |
| `FROM` | What table contains them? |
| `WHERE` | Which rows should remain? |
| `ORDER BY` | How should the result be sorted? |
| `LIMIT` | How many rows should be returned? |

## 3. Keywords, identifiers, values, and operators

Consider:

```sql
WHERE systolic >= 140
```

Identify the parts:

- **Identifier / column:** `systolic`
- **Operator:** `>=`
- **Value:** `140`

Now consider:

```sql
WHERE indication = 'Hypertension'
```

- **Identifier / column:** `indication`
- **Operator:** `=`
- **Value:** `'Hypertension'`

### Important
Numeric values are usually not quoted:

```sql
WHERE systolic >= 140
```

Text values use single quotes:

```sql
WHERE indication = 'Hypertension'
```

````python
# Optional notebook self-check.
# This cell is Python, not SQL. Run it here in Jupyter.

questions = [
    ("In WHERE systolic >= 140, what is the column?", "systolic"),
    ("In WHERE systolic >= 140, what is the operator?", ">="),
    ("In WHERE indication = 'Hypertension', what is the text value?", "Hypertension"),
]

score = 0
for prompt, answer in questions:
    response = input(prompt + " ")
    if response.strip().lower() == answer.lower():
        print("✓ Correct")
        score += 1
    else:
        print(f"✗ Review this one. Expected: {answer}")

print(f"\nScore: {score}/{len(questions)}")
````

## 7. Comparison operators

| Operator | Meaning |
|---|---|
| `=` | equal to |
| `<>` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

### Experiment
Start with:

```sql

FROM bp_readings
WHERE systolic >= 140;
```

Now change **only the operator** and rerun the query several times:

```sql

WHERE systolic < 140
WHERE systolic <= 140
WHERE systolic <> 140
```

## 8. `AND` and `OR` — combine conditions

### Query A

```sql
SELECT patient_id, systolic, diastolic
FROM bp_readings
WHERE systolic >= 140
  AND diastolic >= 90;
```

### Query B

```sql
SELECT patient_id, systolic, diastolic
FROM bp_readings
WHERE systolic >= 140
   OR diastolic >= 90;
```

### Predict before running
Which query will usually return more rows?

- `AND` means ______________________________
- `OR` means _______________________________

Now run both queries and compare the results.

````python
# Optional logic check in the notebook.
# This is Python demonstrating the same Boolean idea.

examples = [
    {"systolic": 150, "diastolic": 92},
    {"systolic": 150, "diastolic": 82},
    {"systolic": 128, "diastolic": 94},
    {"systolic": 128, "diastolic": 78},
]

for x in examples:
    and_result = x["systolic"] >= 140 and x["diastolic"] >= 90
    or_result = x["systolic"] >= 140 or x["diastolic"] >= 90
    print(x, "AND:", and_result, "OR:", or_result)
````

## 11. SQL punctuation and formatting

Read this query carefully:

```sql
SELECT patient_id, reading_date, systolic
FROM bp_readings
WHERE source = 'home'
ORDER BY reading_date;
```

### What the punctuation means
- Commas separate columns.
- Single quotes surround text values.
- A semicolon ends the statement.
- Line breaks make queries easier to read.
- SQL keywords are commonly written in uppercase.

SQL keywords are generally case-insensitive:

```sql
select *
from patients;
```

and:

```sql
SELECT *
FROM patients;
```

mean the same thing.

## 14. Build a query one clause at a time

Instead of writing a long query all at once, build it incrementally.

### Step 1

```sql
SELECT patient_id, systolic, diastolic
FROM bp_readings;
```

Run it.

### Step 2 — add a filter

```sql
SELECT patient_id, systolic, diastolic
FROM bp_readings
WHERE systolic >= 140;
```

Run it.

### Step 3 — add sorting

```sql
SELECT patient_id, systolic, diastolic
FROM bp_readings
WHERE systolic >= 140
ORDER BY systolic DESC;
```

Run it.

### Step 4 — add a limit

```sql
SELECT patient_id, systolic, diastolic
FROM bp_readings
WHERE systolic >= 140
ORDER BY systolic DESC
LIMIT 5;
```

Run it.

## 17. The beginner's SQL pattern
