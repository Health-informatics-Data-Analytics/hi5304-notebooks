## 8. UPDATE — modify an existing value

First inspect the target:
```sql
SELECT * FROM patients WHERE patient_id = 1006;
```

Then update:
```sql
UPDATE patients
SET zip_code = '75206'
WHERE patient_id = 1006;
```

Verify again:
```sql
SELECT * FROM patients WHERE patient_id = 1006;
```

> **Critical warning:** What would happen if you omitted `WHERE patient_id = 1006`?

## 9. UPDATE several variables at once

Suppose the BP record was corrected:
```sql

    source = 'Home'

```

Verify:
```sql
SELECT * FROM bp_readings WHERE bp_id = 10;
```

## 10. ALTER TABLE — modify the structure

Add a new variable:
```sql
ALTER TABLE patients
ADD COLUMN preferred_language TEXT;
```

Inspect the structure:
```sql
\d patients
```

Now add data to the new column:
```sql
UPDATE patients
SET preferred_language = 'English'
WHERE patient_id = 1006;
```

### Key distinction
- `ALTER TABLE` changes the **structure**.
- `UPDATE` changes the **data**.
