Locate the MIMIC-IV Demo files

```python
base = Path("/workspaces/hi5304-notebooks/data/mimic-iv-demo")
hosp, icu = base / "hosp", base / "icu"

def read_mimic(folder, name):
    for ext in (".csv", ".csv.gz"):
        path = folder / (name + ext)
        if path.exists():
            return pd.read_csv(path, low_memory=False)
    raise FileNotFoundError(f"Missing {name} in {folder}")
```

New this week: if / elif / else


```python
def label_los(days):
    if pd.isna(days):
        return "Missing"
    elif days < 3:
        return "Short"
    elif days <= 7:
        return "Moderate"
    else:
        return "Long"
```

Create length of stay safely

```python
admissions["admittime"] = pd.to_datetime(
    admissions["admittime"], errors="coerce")
admissions["dischtime"] = pd.to_datetime(
    admissions["dischtime"], errors="coerce")

admissions["los_days"] = (
    admissions["dischtime"] - admissions["admittime"]
).dt.total_seconds() / 86400
```

Apply a conditional rule

```python
admissions["los_group"] = admissions["los_days"].apply(label_los)

print(admissions["los_group"].value_counts(dropna=False))

# Quality check
print("Negative LOS:", (admissions["los_days"] < 0).sum())
```

Challenge: Turn a loop into a function

```python
def missing_report(df, columns):
    for col in columns:
        n = df[col].isna().sum()
        pct = 100 * n / len(df) if len(df) else 0
        print(f"{col}: {n} missing ({pct:.1f}%)")

missing_report(admissions, ["insurance", "admission_type"])
```

SQLAlchemy: Python talks to PostgreSQL

```python
import os
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

# Create the PostgreSQL connection URL
url = URL.create(
    "postgresql+psycopg",
    username=os.getenv("PGUSER", "hi5304_user"),
    password=os.getenv("PGPASSWORD", "hi5304"),
    host=os.getenv("PGHOST", "db"),
    port=int(os.getenv("PGPORT", "5432")),
    database=os.getenv("PGDATABASE", "hi5304")
)

# Create SQLAlchemy engine
engine = create_engine(url, pool_pre_ping=True)

print("SQLAlchemy engine created successfully.")
```

```python
try:
    with engine.connect() as connection:
        result = connection.execute(text("SELECT version();"))
        print("Connected to PostgreSQL!")
        print(result.scalar())
except Exception as e:
    print("Connection failed:")
    print(e)
```

One question, two languages

```python
import pandas as pd
from sqlalchemy import text

with engine.connect() as conn:
    database_info = pd.read_sql(
        text("""
            SELECT
                current_database() AS database_name,
                current_schema() AS current_schema,
                current_user AS username;
        """),
        conn
    )

display(database_info)

with engine.connect() as conn:
    schemas = pd.read_sql(
        text("""
            SELECT schema_name
            FROM information_schema.schemata
            ORDER BY schema_name;
        """),
        conn
    )

display(schemas)
```
