12. Load our course CSV datasets

```python
from pathlib import Path
import pandas as pd
import numpy as np

# Course data directory
data_dir = Path("/workspaces/hi5304-notebooks/data")

# Load existing datasets
patients = pd.read_csv(data_dir / "patients.csv")
medications = pd.read_csv(data_dir / "medications.csv")

# Display actual columns
print("Patients columns:", patients.columns.tolist())
print("Medications columns:", medications.columns.tolist())

# Create synthetic BP readings for teaching
rng = np.random.default_rng(42)

# Use patient IDs from the existing patient dataset
patient_ids = patients["patient_id"].dropna().unique()

BP_readings = pd.DataFrame({
    "patient_id": rng.choice(patient_ids, size=100, replace=True),
    "systolic": rng.integers(105, 175, size=100),
    "diastolic": rng.integers(60, 105, size=100)
})

print("\nDatasets loaded successfully!")
print("Patients:", patients.shape)
print("Medications:", medications.shape)
print("BP readings:", BP_readings.shape)

display(BP_readings.head())
```

```python
BP_readings.to_csv(
    data_dir / "BP_readings.csv",
    index=False
)

print("Saved:", data_dir / "BP_readings.csv")
```

```python
print("Patients:", len(patients))
print("Medications:", len(medications))
print("BP readings:", len(BP_readings))

print("\nMissing values:")
print(BP_readings.isna().sum())

print("\nBP summary:")
display(BP_readings[["systolic", "diastolic"]].describe())

print("\nUnique patients with BP readings:")
print(BP_readings["patient_id"].nunique())
```

14. Calculations with pandas

```python
BP_readings["pulse_pressure"] = (
    BP_readings["systolic"]
    - BP_readings["diastolic"]
)

print(BP_readings["pulse_pressure"].mean())
```

15. Conditional columns with pandas

```python
BP_readings["review_flag"] = (
    BP_readings["systolic"] >= 140
)

print(BP_readings["review_flag"].value_counts())
```

16. Summarize by patient

```python
summary = (BP_readings
    .groupby("patient_id")
    .agg(mean_sbp=("systolic", "mean"),
         n_readings=("systolic", "count"))
    .reset_index()
)

print(summary.head())
```

17. Join patients to BP summaries

```python
patient_summary = patients.merge(
    summary,
    on="patient_id",
    how="left",
    validate="one_to_one"
)

print(patient_summary.head())
```
