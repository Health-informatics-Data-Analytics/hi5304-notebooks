## 2. Anatomy of a Dataset

- **Variable:** A characteristic that can differ across observations.
- **Observation:** One unit represented in the dataset.
- **Record/row:** A stored set of values for one observation.
- **Population:** The full group of interest.
- **Sample:** The subset actually analyzed.
- **Unit of analysis:** The entity represented by one row.

| Dataset | One row represents | Unit of analysis |
|---|---|---|
| Patient registry | One patient | Patient |
| Admissions file | One hospitalization | Admission |
| Laboratory file | One result | Laboratory result |
| Wearable file | One person-day | Person-day |
| County health file | One county | County |

## 4. Roles Variables Play

- **Outcome:** What we are trying to explain or predict.
- **Predictor:** A variable used to explain or predict the outcome.
- **Exposure:** A factor whose relationship with an outcome is evaluated.
- **Confounder:** A third variable related to both exposure and outcome.
- **Mediator:** A variable through which an exposure may affect an outcome.
- **Moderator:** A variable that changes the strength or direction of a relationship.

Example question: Among adults with hypertension, is physical activity associated with lower systolic blood pressure?

- Exposure: physical activity
- Outcome: systolic blood pressure
- Possible confounders: age, medication use, BMI, diet
- Possible mediator: weight loss

# Part II — Asking Good Questions

## 5. Types of Analytical Questions

- **Descriptive:** What happened?
- **Comparative:** How do groups differ?
- **Associational:** Are variables related?
- **Predictive:** What is likely to happen?
- **Causal:** What would happen if we changed something?
- **Operational/QI:** How can a process be improved?

## 6. Turning a Topic into an Answerable Question

A useful question specifies the population, exposure or intervention, comparison, outcome, time, and setting when applicable.

Example:

> Among adults aged 40–75 hospitalized for heart failure, is diabetes associated with 30-day readmission compared with no diabetes during calendar year 2026?

# Part III — Study Designs

## 7. Common Study Designs

- **Cross-sectional:** measures variables at one point or period.
- **Cohort:** follows or reconstructs a group over time.
- **Case-control:** begins with outcome status and looks backward.
- **Randomized controlled trial:** randomly assigns an intervention.
- **Quasi-experimental:** evaluates an intervention without full randomization.
- **Secondary data analysis:** uses data collected for another purpose.
- **Qualitative:** explores experience and meaning.
- **Mixed methods:** integrates quantitative and qualitative approaches.

## Data Analysis Toolbox #2 — Study Design

Ask:

# My Professional Perspective
## Technology Is Never the Starting Point

Across clinical care, research, education, digital health, and software development, stakeholders rarely begin by asking for a particular technology. They ask why something happened, who is at risk, what should change, or whether an intervention worked.

The software may change—from paper records and spreadsheets to SQL, Python, cloud platforms, and artificial intelligence—but the essential discipline remains the same: understand the healthcare problem, define the population, identify the correct variables, examine data quality, and communicate what the evidence can reasonably support.

> **Professional lesson:** Begin with the decision that needs to be made. Use technology only after the question and analytical purpose are clear.

### Guided Step 5 — Inspect Category Values

```python
categorical_columns = [
    "sex", "admission_type", "diabetes", "smoking_status",
    "readmitted_30d", "discharge_disposition", "study_group"
]

for column in categorical_columns:
    print(f"\n{column}")
    print(data[column].value_counts(dropna=False))
```

### Guided Step 6 — Search for Implausible Numerical Values

```python
numeric_checks = {
    "age_years": (0, 120),
    "systolic_bp_mmHg": (50, 260),
    "ldl_mg_dL": (10, 400),
    "pain_score_0_10": (0, 10),
    "length_of_stay_days": (0, 365),
}

for column, (lower, upper) in numeric_checks.items():
    invalid = data.loc[
        ~data[column].between(lower, upper) & data[column].notna(),
        [column]
    ]
    print(f"{column}: {len(invalid)} potentially invalid value(s)")
    if not invalid.empty:
        display(invalid)
```
