## Why This Matters

Healthcare data are created through clinical care, billing, laboratory testing, operations, surveys, wearable devices, and public health reporting. These systems were not always designed for analysis. Missing values, duplicate records, inconsistent categories, and implausible measurements can distort findings.

Data quality assessment is therefore not housekeeping. It is a professional responsibility.

## Part 2 – Seven Dimensions of Data Quality

| Dimension | Guiding question |
|---|---|
| Completeness | Are required values present? |
| Accuracy | Do recorded values reflect reality? |
| Consistency | Are categories and formats used uniformly? |
| Validity | Do values meet defined rules and ranges? |
| Timeliness | Are data current enough for the decision? |
| Uniqueness | Are records duplicated? |
| Relevance | Do the available variables answer the question? |

A value can be **complete** but still inaccurate or invalid. For example, an age of 214 is present, but it is not plausible.

## Data Analysis Toolbox

Before analyzing a new dataset, ask:

- Who collected the data, and why?
- What does one row represent?
- What does each column represent?
- Is a data dictionary available?
- Are important values missing?
- Are duplicate records present?
- Are formats and units consistent?
- Are values clinically plausible?
- Is the dataset relevant and timely for the question?
